```python
import pandas as pd
import numpy as np

data = {
    "Name":["Ram",'Sita','Hari','Gita','Ram'],
    "Age": [ 20,21,np.nan,20,20],
    "Marks":[85,90,75,np.nan,85]
}
df = pd.DataFrame(data)
df
```

```python
# find the missing values
print(df.isna().sum())
print("Drop missing values:", df.isna().sum().sum())
```

```python
#remove the duplicate value
print(df.duplicated())
```

```python
df_cleaned = df.drop_duplicates()
```

```python
print(df_cleaned.duplicated().sum())
```

```python
# handling missing values:
df_cleaned['Age'] = df_cleaned['Age'].fillna(df_cleaned['Age'].median())
# for mode: .mode()[0]
```

```python
#Outlier
Q1 = df.Marks.quantile(0.25)
Q3 = df.Marks.quantile(0.75)


IQR = Q3 -Q1

#outlier boundaries:
Lower = Q1 - 1.5*(IQR)
Upper = Q3 + 1.5*(IQR)

outliers = df[(df.Marks < Lower) | (df.Marks > Upper)]
outliers
```
