```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from collections import Counter
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix,classification_report, balanced_accuracy_score
```

```python
X,y = make_classification( n_samples=5000,n_features=10, n_informative=5,n_redundant=2,n_classes=2,
                         weights=[0.95,0.05], random_state=42)
print(Counter(y))
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,stratify=y)
```

```python
model = LogisticRegression(max_iter=1000)
model.fit(X_train,y_train)
y_pred = model.predict(X_test)

print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
print("Balance Accuracy:", balanced_accuracy_score(y_test,y_pred))
```

```python
# random sampling :
from imblearn.over_sampling import RandomOverSampler
oversampler = RandomOverSampler(random_state=42)
X_over,y_over = oversampler.fit_resample(X_train,y_train)
print(Counter(y_over))
```

```python
pd.DataFrame(y_over).duplicated().sum()
```

```python
# we can do experiment with class weights and undersampling too.

#SMOTE
from imblearn.over_sampling import SMOTE
smote = SMOTE(random_state=42)
X_smote,y_smote = smote.fit_resample(X_train,y_train)
print(Counter(y_smote))
```
