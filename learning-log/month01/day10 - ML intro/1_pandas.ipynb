```python
import pandas as pd

data= {
	'program_size_kb': [100,200,300,400,500],
	'input_count':[100,200,300,400,500],
	'cpu_threads': [1,1,2,2,4],
	'processor_type': ['ARM','ARM','x86','x86','ARM'],
	'execution_time_ms':[12,24,19,26,18]
}

df = pd.DataFrame(data)
print(df)
```

```python
# separate the features and label
X = df.drop(columns=['execution_time_ms'])
y= df['execution_time_ms']
print("Features: \n",X)
print("Target: \n",y)
```

```python
print(X.shape)
print(y.shape)
```

```python
# identify numerical and categorical features:
numerical_features= X.select_dtypes(include=['number']).columns
categorical_features = X.select_dtypes(include=['str']).columns
print("Numerical Features:", list(numerical_features))
print("Categorical Features:", list(categorical_features))
```

```python
 # Assignment::
dataset={
    "CPU cores": [4,8,2,12,6,4,16,2,8,4,10,6],
    "RAM(GB)":[8,16,4,32,8,8,64,6,32,16,24,12],
    "CPU frequency (GHz)": [2.4,3.2,1.8,3.8,2.6,2.4,4.2,2.0,3.5,2.8,3.0,2.2],
    "Device architecture":["ARM","x86_64","ARM","x86_64","ARM","ARM","x86_64","ARM","x86_64","ARM","x86_64","ARM"],
    "Power consumption(W)":[5.5,65,4.2,125,8.0,7.5,170,3.8,95,6.0,105,9.0]
}
data = pd.DataFrame(dataset)
X = data.drop(columns=['Power consumption(W)'])
y=data['Power consumption(W)']
print("X: \n",X)
print("y: \n", y)
print("How many samples do you have? ", data.shape[0])
print("How many features do you have?", len(X.columns))
print("Separating the numerical and categorical features:")
numerical_features = X.select_dtypes(include=['number'])
categorical_features = X.select_dtypes(include=['str'])
```

```python
Note:<br>
Dataset: the collection of samples used to train, test and evalute the ML model <br>
Sample: an individual rows of the data <br>
Feature: input/ the variables in which the model depends <br>
Target/Label: ouput /result/ the variable which is predicted by the model <br>
Numerical feature: the feature whose values are in number (int/float)<br>
Categorical Feature: the feature whose values are in object/ string format. <br>
We separate X and y because the X contains the model input, while y contains the correct target used during training and testing<br>
If we want to predict RAM usage, Power Consumption (W) column become the target. <br>
```
