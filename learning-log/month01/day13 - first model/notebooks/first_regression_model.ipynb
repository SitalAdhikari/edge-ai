```python
import numpy as np
import pandas as pd

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
load_data = load_diabetes()
```

```python
X = load_data.data
y= load_data.target
```

```python
print(X.shape)
print(y.shape)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3, random_state=10)
```

```python
print(len(X_train))
print(len(X_test))
```

```python
model= LinearRegression()
model.fit(X_train,y_train)
y_pred= model.predict(X_test)
```

```python
print("Let's see first 10 actual and predicted values: \n")
for actual,predicted in zip(y_test[0:10], y_pred[0:10]):
    print(f" Actual: {actual: .2f} , Predicted: {predicted: .2f}")
```

```python
# learned parameters: y = m1.x1+ m2.x2 +m3.x3......+m10.x10+c
print("Coefficent(m):", model.coef_)
print("Intercept(c):", model.intercept_)
```
